## How do we learn?

- Learning is an essential ability of any intelligent agent
- There are two general approaches to learning _deductive_ and _inductive_
    + **Deductive approach** corresponds to learning from a theory, e.g. from a textbook, when an agent gets the knowledge directly.
<div class="mycomment">
In AI this type of learning is used when we encode knowledge of an agent using some knowledge representation language and then use automatic reasoning to make decisions. Also usual programming is somewhat similar to the deductive type of learning, when a programmer encodes her knowledge about the domain in form of computational procedures. However, such agents cannot easily improve their performance over time.
</div>
    + **Inductive learning**, also known as _learning from experience_, is often used in practical settings, when an agent observes the environment and tries to find a model explaining how it works. The obtained knowledge is then used to improve performance of the agent in similar situations.

<div class="myalert"> <b>Machine learning</b> focuses on inductive learning and, therefore, agents using it can improve themselves with experience</div>

## Symbolic and Subsymbolic AI

    - Deductive acquisition, learning at school

# Learning task

ML studies algorithms that
- improve their **performance** $P$
- at some **task** $T$
- given **experience** $E$

Well-defined **learning task** is a triple $\langle T,P,E \rangle$.

<div class="myalert">
    Specify a learning task at the beginning of every ML project!
</div>

## Sample learning tasks

- *Design an agent that plays chess*
    - $T$: play Chess - for every position on a board find the best possible move, which maximizes the expected revenue of an agent,
    - $P$: win the majority of games played in a round,
    - $E$: games played as sequences of moves - human-human, human-agent, agent-agent

- *A continuous integration system must decide if tests must be run on the last push of changes or not*  
    - $T$: predict if code modifications represent changes in program functionality and therefore must be tested,
    - $P$: minimize the number of bugs missed due to skipped tests,
    - $E$: changesets from the repository with labels indicating if a test can be skipped or not

## Design a learning task

- Think about a project using machine learning and write its short and informal description
- Afterwards, use the description to specify the learning task as precise as possible:
    - Define the training experience
    - Determine type of knowledge to be learned and define the target function
    - Find a good representation for your function
    - Select a learning algorithm to fit your target function to training experience
- Describe the trade-offs/simplifications/abstractions you considered while defining your learning task

### Experience

+ Analyze which data can be used as training experience
+ Type of feedback
    - **direct**: every message in the inbox is labeled as *spam* or *not*, vs
    - **indirect**: we know labels of messages and the overall user satisfaction score with the spam filter
+ Control over sequences of examples
    - a user can select representative messages and provide labels
    - learner selects the most problematic cases - **active learning**
    - random process
+ How well the data represents the real-world
    - Data states that all horses are black
+ And many more: sources, volumes, noise, trust, relevancy, etc.
+ Do not forget to assess costs: getting data can be expensive

### Performance (classification case)

- Comparison of predicted and actual labels for binary classification problem results in four values that form a **confusion matrix**
  - $\mathit{TP}$ - number of true positives, "true" labels that were predicted as such
  - $\mathit{FP}$ - number of false positives, type I error, rejecting a correct hypothesis
  - $\mathit{TN}$ - number of true negatives
  - $\mathit{FN}$ - number of false negatives, type II error, non-rejection of a false hypothesis
  
$$
\begin{array}{|c|c|}\hline
    \mathit{TP} & \mathit{FP} \\
    \hline
    \mathit{FN} & \mathit{TN} \\ \hline
\end{array}
$$

- There are various performance measures which are selected wrt. particular learning task
- Accuracy, good for balanced data
  
\begin{equation*}
 A = \frac{\mathit{TP} + \mathit{TN}} {\mathit{total}}
\end{equation*}

  - Precision, recall, and $F_1$-score, work for imbalanced data
  
\begin{equation*}
 P = \frac{\mathit{TP}}{\mathit{TP} + \mathit{FP}} ~~~~~
 R = \frac{\mathit{TP}}{\mathit{TP} + \mathit{FN}} ~~~~~
 F_1 = \frac{2PR}{P+R}
\end{equation*}

See, e.g., *Marina Sokolova, Guy Lapalme: [A systematic analysis of performance measures for classification tasks.](http://rali.iro.umontreal.ca/rali/sites/default/files/publis/SokolovaLapalme-JIPM09.pdf) Inf. Process. Manag. 45(4): 427-437 (2009)* for an overview

Metrics usually shipped with ML frameworks, e.g., [Sklearn](https://scikit-learn.org/stable/modules/classes.html#module-sklearn.metrics) or [Torch](https://torchmetrics.readthedocs.io/en/latest/)

### Define an ML problem

- **Classification**: given a set of objects
    - Binary or multi-class classification:
        - partition objects into $k$ categories
        - learn a probability distribution over mutually disjoint classes
    - Multi-label classification:
        - assign objects to at least one of the $k$ categories
        - learn a marginal probability distribution
- **Regression**: given some input predict a numerical value
- **Clustering**: group similar observations in clusters

## Determine the model

+ Defines the exact type of knowledge to be learned
    - Target function $f : I \to O$ mapping an input to some output
    - Example: given a student profile, output possible study length
    - Example: given a set of documents, output a set of groups
      comprising similar documents
+ Learning some functions might be difficult!
    - Policy learning: $\mathit{BestMove}: \mathit{Position} \to \mathit{Move}$ (often hard)  
    - Action-value learning: $\mathit{Quality}: \mathit{Position} \to \mathbb{R}$ (relatively easy)

## Knowledge representation

+ A model provides a representation for the target knowledge
+ Model should be able to capture the theory you want to learn
    - Possibilities: sets of logic rules, linear functions, (linear) combinations
      of functions, sets (matrices) of representative data points, etc.
    - Example: every board in Chess gets a quality value (a large matrix)
    - Example: the revenue of a company depends linearly on a size of the advertisement budget

<div class="myalert">
    A learner that makes <b>no prior assumptions</b> regarding the identity of the target concept has <b>no rational basis</b> for classifying any unseen instances (T. Mitchell, Machine Learning)
</div>

### Data preparation

- Given a well-defined training experience
- **Open questions**:
  + How do we *evaluate the performance measure* to check the quality?
     <div class="mycomment">
    Fit a model on a <i>training dataset</i> and then use the model to predict labels of observations in a <i>test dataset</i>. Compare predictions with real labels using some performance measure.
    </div>

  + Given only a small number of observations, where can we get the ones for testing?
  <div class="mycomment">
       Split existing ones into training and test sets with a ratio of, e.g., 75% to 25%
    </div>

  + How should we sample?
  <div class="mycomment">
      There are many possible algorithms allowing for a random selection of observations from the dataset. Different samples might result in different models and, consequently, influence classification performance of a model in the future. It is important to select an appropriate sampling algorithm!
    </div>

## Training algorithm
+ Selection/design of a learning algorithm depends on a model
+ Must theoretically be able to fit a model to data: soundness, completeness
+ Computationally efficient - practical applicability (complexity)
+ Satisfy you requirements for precision and recall: optimization of the performance measure

## Validation

- Quite often learning algorithms can be parametrized
- Selection of best values for **hyperparameters** is complicated
  - How can we avoid **overfitting**, i.e., check how well a model generalizes to new data?
- Validation of models trained with different setting is done using a **validation** set, which can be
   - a specifically designed set of observations
   - sampled from existing data - k-fold cross validation

### k-fold Cross-Validation
- Shuffle the dataset
- Partition the dataset into k equal groups, and for each group
    - Set the group as the test dataset
    - Union remaining groups into a training dataset
    - Fit a model and evaluate it using these sets
    - Store the evaluation score
- Estimate the performance of the model using stored evaluation scores
