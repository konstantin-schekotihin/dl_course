### How Do We Learn?

- Learning is an essential ability of any intelligent agent
- There are two general approaches to learning _deductive_ and _inductive_
    + **Deductive approach** corresponds to learning from a theory, e.g. from a textbook, when an agent gets the knowledge directly.
    + **Inductive learning**, also known as _learning from experience_, is often used in practical settings, when an agent observes the environment and tries to find a model explaining how it works. The obtained knowledge is then used to improve performance of the agent in similar situations.

> 📝 **Note:** In AI, deductive reasoning encodes domain expertise using knowledge representation languages and executes automated deduction (similar to algorithmic software development). However, such systems cannot easily adapt or improve performance autonomously over time.

> 💡 **Key Takeaway:** **Machine learning** is fundamentally **inductive inference**: computational algorithms that extract generalizable models directly from empirical experience.

## 3. The Formal Learning Task: $(T, P, E)$

ML studies algorithms that
- improve their **performance** $P$
- at some **task** $T$
- given **experience** $E$

Well-defined **learning task** is a triple $(T, P, E)$.

> ⚠️ **Important:** Always clearly specify the learning task $(T, P, E)$ at the beginning of every machine learning project.

### Benchmark Learning Tasks

- *Design an agent that plays chess*
    - $T$: play Chess - for every position on a board find the best possible move, which maximizes the expected revenue of an agent,
    - $P$: win the majority of games played in a round,
    - $E$: games played as sequences of moves - human-human, human-agent, agent-agent

- *A continuous integration system must decide if tests must be run on the last push of changes or not*  
    - $T$: predict if code modifications represent changes in program functionality and therefore must be tested,
    - $P$: minimize the number of bugs missed due to skipped tests,
    - $E$: changesets from the repository with labels indicating if a test can be skipped or not

### Task ($T$): Defining the Problem

- **Classification**: given a set of objects
    - Binary or multi-class classification:
        - partition objects into $K$ classes
        - learn a probability distribution over mutually disjoint classes
    - Multi-label classification:
        - assign objects to at least one of the $K$ classes
        - learn a marginal probability distribution
- **Regression**: given some input predict a numerical value
- **Clustering**: group similar observations in clusters

### Performance Metric ($P$): The Confusion Matrix

- Evaluating a binary classifier on $N$ labeled observations yields four fundamental counts forming a **confusion matrix**:
  - $\mathrm{TP}$ (True Positive): Actual positives correctly predicted as positive.
  - $\mathrm{FP}$ (False Positive): Actual negatives incorrectly predicted as positive (Type I error).
  - $\mathrm{FN}$ (False Negative): Actual positives incorrectly predicted as negative (Type II error).
  - $\mathrm{TN}$ (True Negative): Actual negatives correctly predicted as negative.

$$
\begin{array}{c|cc}
& \hat{t} = 1 \text{ (Predicted)} & \hat{t} = 0 \text{ (Predicted)} \\
\hline
t = 1 \text{ (Actual)} & \mathrm{TP} & \mathrm{FN} \\
t = 0 \text{ (Actual)} & \mathrm{FP} & \mathrm{TN}
\end{array}
$$

*(where $N = \mathrm{TP} + \mathrm{TN} + \mathrm{FP} + \mathrm{FN}$ is the total evaluated sample count).*


- Performance metrics are computed directly from the confusion matrix entries:
- **Accuracy** (fraction of correct predictions across all $N$ samples; best for balanced classes):

$$
\mathrm{Accuracy} = \frac{\mathrm{TP} + \mathrm{TN}}{N} = \frac{\mathrm{TP} + \mathrm{TN}}{\mathrm{TP} + \mathrm{TN} + \mathrm{FP} + \mathrm{FN}}
$$

- **Precision, Recall, and $F_1$-score** (critical for imbalanced class distributions):
  - **Precision**: Fraction of predicted positives that are genuinely positive.
  - **Recall**: Fraction of actual positives that were successfully retrieved (True Positive Rate).
  - **$F_1$-score**: Harmonic mean balancing precision and recall.

$$
\mathrm{Precision} = \frac{\mathrm{TP}}{\mathrm{TP} + \mathrm{FP}}, \quad \mathrm{Recall} = \frac{\mathrm{TP}}{\mathrm{TP} + \mathrm{FN}}, \quad F_1 = \frac{2 \cdot \mathrm{Precision} \cdot \mathrm{Recall}}{\mathrm{Precision} + \mathrm{Recall}}
$$

- **Specificity**: fraction of actual negatives correctly identified, $\mathrm{Specificity} = \frac{\mathrm{TN}}{\mathrm{TN} + \mathrm{FP}}$.
- Synonyms you will meet: recall = **sensitivity** = true positive rate; precision = positive predictive value.

<small>📖 *Source: [Sokolova, M., & Lapalme, G. (2009). A systematic analysis of performance measures for classification tasks. Information Processing & Management, 45(4), 427–437](https://doi.org/10.1016/j.ipm.2009.03.002)*</small>

Metrics usually shipped with ML frameworks, e.g., [Scikit-Learn](https://scikit-learn.org/stable/modules/classes.html#module-sklearn.metrics) or [PyTorch](https://torchmetrics.readthedocs.io/en/latest/).

### Performance Visualization: The ROC Curve & AUC

- **Receiver Operating Characteristic (ROC)**: evaluates a classifier across all decision thresholds $\tau$ on its score, which can be a probability $p(t=1 \mid \mathbf{x})$ or any other score (e.g. of an SVM); the classifier predicts $\hat{t} = 1$ when the score is $\ge \tau$:
  - **True Positive Rate ($\mathrm{TPR}$)**: $\mathrm{TPR} = \frac{\mathrm{TP}}{\mathrm{TP} + \mathrm{FN}}$ (identical to recall / sensitivity).
  - **False Positive Rate ($\mathrm{FPR}$)**: $\mathrm{FPR} = \frac{\mathrm{FP}}{\mathrm{FP} + \mathrm{TN}} = 1 - \mathrm{Specificity}$ (fraction of negatives misclassified).

<center>
    <img src="https://upload.wikimedia.org/wikipedia/commons/1/13/Roc_curve.svg" width="28%"/>
</center>

- **Area Under the Curve (AUC)**, computed on held-out data: $0.5$ = no better than chance, $1.0$ = perfect; a good ROC curve hugs the top-left corner.

<small>📖 *Figure: [Wikipedia: Receiver Operating Characteristic](https://en.wikipedia.org/wiki/Receiver_operating_characteristic)*</small>

## 5. Statistical Learning: Choosing a Model

### The Target Function

+ Defines the exact type of knowledge to be learned
    - Target function $f : I \to O$ mapping an input to some output
    - Example: given a student profile, output possible study length
    - Example: given a set of documents, output a set of groups
      comprising similar documents
+ Learning some functions might be difficult!
    - Policy learning: $\mathit{BestMove}: \mathit{Position} \to \mathit{Move}$ (often hard)  
    - Action-value learning: $\mathit{Quality}: \mathit{Position} \to \mathbb{R}$ (relatively easy)

### What Is Statistical Learning?

- We observe a target $t$ and features $\mathbf{x} = (x_1, \dots, x_D)^\mathrm{T}$ and assume that they are related by

$$
t = f(\mathbf{x}) + \epsilon
$$

  - $f$ is a fixed but unknown function: the systematic information that $\mathbf{x}$ provides about $t$ (the target function of the previous slide);
  - $\epsilon$ is random noise with mean zero that does not depend on $\mathbf{x}$: unmeasured influences and random variation.
- **Statistical learning** is the set of methods for estimating $f$ from data. The estimate is our model $y(\mathbf{x})$, its prediction is $\hat{t} = y(\mathbf{x})$.
- Example (Lecture 02): sales $t$ of a product in $200$ markets and its TV, radio and newspaper advertising budgets $\mathbf{x}$.

<small>📖 *Source: [James, G., Witten, D., Hastie, T., Tibshirani, R., & Taylor, J. (2023). An Introduction to Statistical Learning with Applications in Python, Ch. 2: Statistical Learning. Springer](https://doi.org/10.1007/978-3-031-38747-0_2)*</small>

### Why Estimate $f$? Prediction vs. Inference

- **Prediction**: only an accurate $\hat{t}$ matters, the model may be a black box.
  - Example: a direct-marketing campaign mails an offer only to customers who are likely to respond.
- **Inference**: we want to understand *how* $t$ is associated with $\mathbf{x}$: which features matter, how strongly, in which direction.
  - Example: which media are associated with sales, and how many more units are sold per additional \$1000 of TV advertising?
- The error of a prediction $\hat{t}$ has two parts:
  - **Reducible error**: $y(\mathbf{x})$ is not a perfect estimate of $f(\mathbf{x})$; better methods and more data reduce it.
  - **Irreducible error**: even with $y(\mathbf{x}) = f(\mathbf{x})$ the noise $\epsilon$ remains. It cannot be predicted from $\mathbf{x}$ and sets a floor that no model can beat.
- Averaged over many observations with the same $\mathbf{x}$, the squared prediction error splits into these two parts:

$$
\text{average of } (t - \hat{t})^2 = \underbrace{\left(f(\mathbf{x}) - y(\mathbf{x})\right)^2}_{\text{reducible}} + \underbrace{\mathrm{var}[\epsilon]}_{\text{irreducible}}
$$

### Knowledge & Feature Representation

+ A model provides a representation for the target knowledge
+ Model should be able to capture the theory you want to learn
    - Possibilities: sets of logic rules, linear functions, (linear) combinations
      of functions, sets (matrices) of representative data points, etc.
    - Example: every board in Chess gets a quality value (a large matrix)
    - Example: a linear function of the TV, radio and newspaper budgets represents the sales (previous slides)
+ The **inductive bias** of a learner: the assumptions about the target function that it makes beyond the data, above all the choice of its representation

> 💡 **The Necessity of Inductive Bias:** A learner that makes **no prior assumptions** regarding the identity of the target concept has **no rational basis** for classifying any unseen instances.
>
> <small>📖 *Source: [Mitchell, T. M. (1997). Machine Learning. McGraw-Hill](https://www.cs.cmu.edu/~tom/mlbook.html)*</small>

### Parametric vs. Non-Parametric Models

- **Parametric** methods assume a form of $f$, e.g. a linear model $y(\mathbf{x}, \mathbf{w}) = w_0 + w_1 x_1 + \dots + w_D x_D$. Learning then means estimating the $D + 1$ weights.
  - ➕ easy to fit and to interpret; works with relatively few observations
  - ➖ if the assumed form is far from the true $f$, the estimate is poor, however much data we have
- **Non-parametric** methods assume no fixed form; the model follows the data, e.g. $k$-nearest neighbors (Lecture 01).
  - ➕ can fit many different shapes of $f$
  - ➖ needs many more observations, and can follow the noise of the training data
- Choosing a parametric form is a strong inductive bias (previous slide).

### Flexibility vs. Interpretability

| Method (lecture) | Flexibility | Interpretability |
| :--- | :---: | :---: |
| Lasso (02) | low | high |
| Linear and logistic regression (02, 03) | low | high |
| Decision trees (05) | medium | high |
| $k$-nearest neighbors (01) | set by $k$ | by similar examples |
| Support vector machines with kernels (06) | high | low |
| Random forests, boosting (05) | high | low |
| Neural networks (04) | high | low |

- Why choose a less flexible method?
  - For **inference** we need a model whose parameters we can read.
  - Even for **prediction**, the most flexible model is often *not* the most accurate on new data: it overfits (next slides).
- *There is no free lunch*: no single method is best on all data sets. Choosing a method for the data at hand is part of the job.

<small>📖 *Source: [James, G., Witten, D., Hastie, T., Tibshirani, R., & Taylor, J. (2023). An Introduction to Statistical Learning with Applications in Python, Ch. 2: Statistical Learning. Springer](https://doi.org/10.1007/978-3-031-38747-0_2)*</small>

### Training Error vs. Generalization Error

- A model is fitted by making its error on the **training data** small, e.g. the mean squared error $\mathrm{MSE} = \frac{1}{N}\sum_{n=1}^{N} (t_n - \hat{t}_n)^2$ or, in classification, the fraction of misclassified observations.
- What we care about is the error on **new data** that the model has not seen (tomorrow's sales, the next patient): the **generalization error**, often also called the *test error*.
- As the flexibility of the model grows,
  - the training error keeps decreasing,
  - the generalization error first decreases, then increases again (a U-shape); it never falls below the irreducible error.

> 📘 **Definition: Overfitting.** A model overfits when its training error is small but its generalization error is large: it follows the noise of its training data, and a less flexible model would predict new data better. It shows as a gap between the training error and the error on **validation** data (§6).

- Lectures 01 and 02 compute these curves on real data ($k$-NN, polynomial regression), with the cross-validation error in place of the unknown generalization error.

### The Bias–Variance Trade-Off

<center>
    <img src="https://raw.githubusercontent.com/konstantin-schekotihin/dl_course/master/shared/images/bias-variance.png" width="32%"/>
</center>

- Imagine fitting the same method on many different training sets from the same population; each dot is the prediction of one fitted model, the centre is the true value:
  - **Bias**: how far the predictions are off *on average*: the error of approximating a complex reality by a too simple model.
  - **Variance**: how much the prediction changes from one training set to the next: flexible models follow their particular sample.
- Generalization error at a given $\mathbf{x}$ = $\mathrm{bias}^2$ + variance + irreducible error.
- More flexibility lowers the bias but raises the variance. The best flexibility lies in between, differs from data set to data set, and must be **estimated** from data (§6).

<small>📖 *Source: [James, G., Witten, D., Hastie, T., Tibshirani, R., & Taylor, J. (2023). An Introduction to Statistical Learning with Applications in Python, Ch. 2: Statistical Learning. Springer](https://doi.org/10.1007/978-3-031-38747-0_2)*</small>

### The Optimization & Training Algorithm
+ Selection/design of a learning algorithm depends on a model
+ Must theoretically be able to fit a model to data: soundness, completeness
+ Computationally efficient - practical applicability (complexity)
+ Satisfy your requirements for precision and recall: optimization of the performance measure

### Estimating the Error on New Data: The Test Set

- Given a well-defined training experience
- **Open questions**:
  + How do we *evaluate the performance measure* to check the quality?
     > 📝 **Note:** Fit a model on a **training dataset** and then use the model to predict labels of observations in an independent **test dataset**. Compare predictions with ground-truth targets using an appropriate performance metric.

  + Given only a small number of observations, where can we get the ones for testing?
  > 📝 **Note:** Partition existing observations into disjoint training and test sets (e.g. 80% / 20% or 75% / 25%). With few observations a single split gives a noisy estimate and leaves less data for training: use $S$-fold cross-validation on the training part (below).

  + How should we sample?
  > 📝 **Note:** When class distributions are skewed, simple random sampling can lead to severe data imbalance across splits. Always use stratified sampling to preserve class proportions.

### Generalization, Overfitting & Validation

- Most learning algorithms have **hyperparameters** that set the flexibility of the model, e.g. $k$ in $k$-NN or the degree of a polynomial.
- Two different jobs:
  - **Model selection**: choose the hyperparameters, i.e. the flexibility.
  - **Model assessment**: estimate how well the final model performs on new data.
- To choose hyperparameters without overfitting, estimate the generalization error on data that was **not used for fitting**:
  - **Validation set approach**: split the training data once into a part for fitting and a **validation** set. Drawbacks: the estimate depends strongly on which observations land in which part, and the model is fitted on fewer observations, so the error tends to be overestimated.
  - **$S$-fold cross-validation** uses every training observation for fitting and for validation (next slide).

### $S$-Fold Cross-Validation

- Gives a test-error estimate that is far less variable than a single validation split. It is slightly pessimistic, because each model is fitted on only $(S - 1)/S$ of the data.
- Randomly partition the training data into $S$ disjoint folds of approximately equal size ($S \ge 2$, typically $S = 5$ or $10$).
- For each fold $s \in \{1, \dots, S\}$:
    - Treat fold $s$ as the hold-out validation set $\mathcal{D}_{\text{val}}$.
    - Train the model on the remaining $S - 1$ folds: $\mathcal{D}_{\text{train}} = \bigcup_{j \neq s} \text{fold}_j$.
    - Compute and record evaluation metric $P_s$ on $\mathcal{D}_{\text{val}}$.
- The **CV estimate** of the performance of this hyperparameter setting is the mean score $\bar{P} = \frac{1}{S}\sum_{s=1}^S P_s$.

| Round | Fold 1 | Fold 2 | Fold 3 | Fold 4 | Fold 5 |
| :---: | :---: | :---: | :---: | :---: | :---: |
| $s = 1$ | **validate** | train | train | train | train |
| $s = 2$ | train | **validate** | train | train | train |
| $s = 3$ | train | train | **validate** | train | train |
| $s = 4$ | train | train | train | **validate** | train |
| $s = 5$ | train | train | train | train | **validate** |

- **Model selection with CV**: compute $\bar{P}$ for every candidate setting, pick the best one, refit the model with it on *all* training data, and report its performance once on the test set.

<small>📖 *Source: [James, G., Witten, D., Hastie, T., Tibshirani, R., & Taylor, J. (2023). An Introduction to Statistical Learning with Applications in Python, Ch. 5: Resampling Methods. Springer](https://doi.org/10.1007/978-3-031-38747-0_5)*</small>

### Summary: Training, Validation and Test Data

| Data | Used to | Never used to |
| :--- | :--- | :--- |
| **Training set** | fit the parameters (weights, tree splits, stored neighbours) | judge the model: the training error is too optimistic |
| **Validation data**: a held-out part of the training data, or the validation folds of CV | choose hyperparameters, features, preprocessing, decision thresholds, the number of epochs, and between models; detect overfitting (training vs. validation error) | report the final performance: it was used for choosing |
| **Test set**: split off before model selection starts | assess the final model **once** | choose or change anything |

> ⚠️ **Important:** As soon as a test score changes a decision (another $k$, feature, threshold, model or random seed), the test set has become validation data and its score overestimates the performance on new data. Lecture 01 measures by how much.
