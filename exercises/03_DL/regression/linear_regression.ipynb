```python
# Required imports - DO NOT CHANGE
from abc import ABC, abstractmethod
```

```python
# Base class - DO NOT CHANGE
class RegressionModel(ABC):
    # Initialise the model weights here
    def __init__(self):
        self.weights = None
    
    # fit() should update the model weights given a training set.
    @abstractmethod
    def fit(self, X, y) -> None:
        pass

    # predict() should return the model's predictions for a given input.
    @abstractmethod
    def predict(self, X):
        pass
```
