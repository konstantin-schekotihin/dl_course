## How Do We Learn?

- Learning is the core hallmark of an intelligent agent.
- There are two general paradigms of learning: **deductive** and **inductive**.
    + **Deductive learning** corresponds to learning from theory, axioms, and logical rules (e.g., studying a textbook or executing programmatic algorithms).
      > 📝 **Note:** In AI, deductive reasoning encodes domain expertise into formal representations (such as knowledge graphs and ontologies) and uses automated theorem provers or search engines to deduce conclusions. While precise, such systems cannot easily adapt or improve performance autonomously from raw data.

    + **Inductive learning**, also known as *learning from experience*, operates in practical settings where an agent observes the environment, collects samples, and infers a general underlying model to make decisions in novel situations.


> 💡 **Key Takeaway:** **Machine learning** is fundamentally **inductive inference**: computational algorithms that extract generalizable models directly from empirical experience.


# Symbolic and Subsymbolic AI

Two distinct paradigms for representing and manipulating knowledge in artificial intelligence.


## Symbolic vs. Subsymbolic Representations

- **Symbolic Representation** (Logic, Ontologies, Graphs):
    - Knowledge is represented explicitly using discrete symbols, predicates, and formal rules:
        - $\textit{patient}_{1}$ has_lesion $\textit{Melanoma}$
        - $\textit{Melanoma} \;\sqsubseteq\; \textit{MalignantNeoplasm}$ (every melanoma is a malignant neoplasm; $\sqsubseteq$ reads "is a subclass of")
    - Inference is conducted via logical deduction, heuristic tree search, and constraint satisfaction.
- **Subsymbolic Representation** (Continuous Functions, Numerical Vectors):
    - Knowledge is encoded implicitly across continuous multi-dimensional parameter arrays (weights $\mathbf{w}$):
        - Linear decision boundary: $\mathbf{w}^\mathrm{T}\mathbf{x} + w_0 > 0$
        - Deep neural network with $L$ layers of weights: $f(\mathbf{x}; \mathbf{W}_1, \dots, \mathbf{W}_L)$

## Comparing AI Paradigms

- **Symbolic AI (GOFAI — Good Old-Fashioned AI)**
    - Deductive knowledge acquisition curated by human domain experts.
    - Exact, provable inference using search and constraint satisfaction.
    - *Bottlenecks*: High brittleness under sensory noise, combinatorial complexity in open-world domains, and manual knowledge engineering bottlenecks.
- **Subsymbolic AI (Connectionist / Machine Learning)**
    - Inductive knowledge acquisition learned directly from sensory data.
    - Continuous numerical optimization and probabilistic inference under uncertainty.
    - *Strengths*: High perceptual robustness to noise, handling rich high-dimensional data (images, audio, text), and end-to-end representation learning.


# A Brief History of Machine Learning

Understanding the historical oscillation between symbolic and neural approaches across three distinct waves.


## Biological Inspiration

<center>
    <img src="https://raw.githubusercontent.com/konstantin-schekotihin/dl_course/master/shared/images/neuron.jpg" width="40%"/>
</center>

<small>📖 *Source: [Russell, S., & Norvig, P. (2020). Artificial Intelligence: A Modern Approach (4th ed.). Pearson](http://aima.cs.berkeley.edu)*</small>

- The human brain has about $9 \times 10^{10}$ neurons, each with several thousand synapses (about $10^{14}$ synapses in total).
- A neuron switches in about $10^{-3}$ s, yet we recognize a scene in about $10^{-1}$ s.
- That leaves time for only about $10^{-1} / 10^{-3} = 100$ sequential steps $\Rightarrow$ the brain must compute massively in parallel.

<small>📖 *Source: [Mitchell, T. M. (1997). Machine Learning, Section 4.1.1. McGraw-Hill](https://www.cs.cmu.edu/~tom/mlbook.html)*</small>

## The Artificial Neuron Model

- McCulloch & Pitts (1943) formalized the first mathematical abstraction of a biological neuron:
    - **Linear combination** of the inputs with weights $w_i$ and a bias $w_0$ (in 1943 the weights and the threshold were fixed by hand; learning them came with the perceptron):
      $$
      a = \sum_{i=1}^D w_i x_i + w_0 = \mathbf{w}^\mathrm{T}\mathbf{x} + w_0
      $$
    - **Nonlinear activation function** producing the output; here the step function
      $$
      y = f(a) = \mathrm{hardlim}(a) = \begin{cases} 1, & a > 0 \\ 0, & a \leq 0 \end{cases}
      $$
    - The neuron "fires" ($y = 1$) when the weighted input exceeds the threshold.

## The Perceptron & The Mark 1 Hardware (1950s–1960s)

- Frank Rosenblatt (1958) created the **Perceptron**, equipping the artificial neuron with adjustable weights and a supervised learning rule.
- Implemented in dedicated analog hardware: the **Mark 1 Perceptron** (1960) at Cornell Aeronautical Laboratory.
    - $20 \times 20$ array of cadmium sulphide photocells (400-pixel camera).
    - Patchboard for flexible input wiring.
    - Racks of motor-driven potentiometers adjusting weights dynamically during training.
- Rosenblatt's perceptron outputs $\pm 1$ with the sign function $\mathrm{sgn}(a) = +1$ if $a \geq 0$ and $-1$ otherwise; this choice also gives the learning rule its simple form *(→ Lecture 05_classification)*. Equivalently, outputs $0/1$ use $\mathrm{hardlim}(a)$ from the previous slide.

## The XOR Crisis & The First AI Winter

- A single-layer perceptron defines a linear hyperplane decision boundary $\mathbf{w}^\mathrm{T}\mathbf{x} + w_0 = 0$.
- In 1969, Marvin Minsky and Seymour Papert published their influential monograph *Perceptrons*, mathematically demonstrating that single-layer perceptrons cannot learn simple non-linearly separable functions such as **XOR** (Exclusive OR).

<center>
    <img src="https://raw.githubusercontent.com/konstantin-schekotihin/dl_course/master/shared/images/ann-linearly-separable.jpg" width="60%"/>
</center>

> ⚠️ **Important:** Minsky and Papert also conjectured, wrongly, that networks with several layers would share these limitations. Together with the lack of any algorithm to train multilayer networks, this contributed to the loss of interest and funding for neural networks during the 1970s and early 1980s.

<small>📖 *Source: [Minsky, M., & Papert, S. (1969). Perceptrons: An Introduction to Computational Geometry. MIT Press](https://mitpress.mit.edu/9780262631112/perceptrons/)*</small>


## Connectionism & Backpropagation (1980s–1990s)

- **Overcoming the linear barrier**: Stacking layers of neurons into **multi-layer feedforward networks (MLPs)** with hidden representations.
- Replacing step functions with smooth, differentiable activations, like sigmoid $\sigma(a) = \frac{1}{1 + e^{-a}}$.
- **The Backpropagation Algorithm**: Using the chain rule of calculus to propagate error gradients backwards from output to hidden units, updating weights via gradient descent.

<small>📖 *Source: [LeCun, Y., Bottou, L., Bengio, Y., & Haffner, P. (1998). Gradient-based learning applied to document recognition. Proceedings of the IEEE, 86(11), 2278–2324](https://doi.org/10.1109/5.726791)*</small>


## Neural Networks Fall Out of Favour (late 1990s–2000s)

- Training networks with many layers in the 1990s stalled due to:
    - **Vanishing and exploding gradients** through deep cascades of saturating activations.
    - **Severe computational bottlenecks**: CPUs took weeks to train even modest architectures.
    - **Severe data scarcity**: Small datasets led to rampant overfitting.
- Research turned to alternatives: **kernel methods** and **support vector machines (SVMs)**, whose training is a convex optimization problem, **Gaussian processes**, and ensemble methods such as random forests.


## The Deep Learning Revolution (2012–Present)

- **The 2012 AlexNet Breakthrough**: Krizhevsky, Sutskever, & Hinton cut the top-5 error on the ImageNet Challenge from 26.2% (best competitor) to 15.3% by training an 8-layer deep convolutional neural network on two NVIDIA GTX 580 GPUs.
- **The Three Pillars of Modern Deep Learning**:
    1. **Big Data**: Massive internet-scale datasets (ImageNet, Common Crawl, LAION).
    2. **Massive Compute**: High-throughput parallel accelerators (GPUs, TPUs, specialized tensor cores).
    3. **Algorithmic Innovations**: Non-saturating activations (ReLU), initialization strategies, residual skip-connections (ResNet), adaptive optimizers (Adam), normalization layers (BatchNorm, LayerNorm), and self-attention (Transformers).

<small>📖 *Source: [Krizhevsky, A., Sutskever, I., & Hinton, G. E. (2012). ImageNet classification with deep convolutional neural networks. NeurIPS](https://proceedings.neurips.cc/paper/2012/hash/c399862d3b9d6b76c8436e924a68c45b-Abstract.html)*</small><br>
<small>📖 *Source: [Deng, J., Dong, W., Socher, R., et al. (2009). ImageNet: A large-scale hierarchical image database. CVPR](https://doi.org/10.1109/CVPR.2009.5206848)*</small>

## Exponential Compute Scaling

- The compute required to train state-of-the-art neural networks has grown exponentially in two phases:
    - until about 2012: doubling every ~2 years, in line with Moore's law;
    - since 2012 (deep learning): doubling every ~3.4 months, about $10\times$ per year.
- Unit: one petaflop/s-day is $10^{15}$ floating point operations per second for 24 hours, about $10^{20}$ operations.

## The Bitter Lesson

> 💡 **Key Takeaway: The "Bitter Lesson" of AI**
>
> *"We have to learn the bitter lesson that building in how we think we think does not work in the long run. The bitter lesson is based on the historical observations that 1) AI researchers have often tried to build knowledge into their agents, 2) this always helps in the short term, and is personally satisfying to the researcher, but 3) in the long run it plateaus and even inhibits further progress, and 4) breakthrough progress eventually arrives by an opposing approach based on scaling computation by search and learning. ## Examples of Core Machine Learning Problem Classes

- **Classification**: Partition objects into categories:
    - *Binary or Multi-class*: Assign each object to one of $K$ mutually disjoint classes $\mathcal{C}_1, \dots, \mathcal{C}_K$; the target is $t \in \{0, 1\}$ for two classes or a 1-of-$K$ vector $\mathbf{t}$.
    - *Multi-label*: Assign objects to arbitrary subsets of classes.
- **Regression**: Given input $\mathbf{x}$, predict continuous numerical target values $t \in \mathbb{R}$.
- **Sequence Learning**: Map sequential inputs to sequential outputs (machine translation, speech recognition, NLP).
- **Representation Learning**: Automatically project high-dimensional inputs (pixels, text) into low-dimensional latent vectors $\mathbf{z} \in \mathbb{R}^M$ with $M < D$, preserving semantic similarity.
- **Generative Modeling**: Estimate underlying probability densities $p(\mathbf{x})$ and sample high-fidelity synthetic data.


# The Formal Learning Framework

How do we formulate an empirical problem mathematically as a machine learning task?


## Mitchell's Learning Task Formulation

> 📘 **Definition: Machine Learning Task**
>
> A computer program is said to **learn** from experience $E$ with respect to some class of tasks $T$ and performance measure $P$, if its performance at tasks in $T$, as measured by $P$, improves with experience $E$.

- Every well-defined machine learning project begins with the formal specification of the triple:
  $$\langle T, P, E \rangle$$
  - **Task ($T$)**: The specific capability or mapping the agent must learn (e.g. mapping inputs $\mathbf{x}$ to targets $t$).
  - **Performance Measure ($P$)**: The quantitative objective metric evaluating the quality of predictions.
  - **Experience ($E$)**: The observational data or environment interaction feedback available to the learner.

<small>📖 — *[Mitchell, T. M. (1997). Machine Learning. McGraw-Hill](https://www.cs.cmu.edu/~tom/mlbook.html)*</small>


# Deep Learning Milestones as $\langle T, P, E \rangle$ Case Studies

Let us consider some illustrative examples of the $\langle T, P, E \rangle$ learning task.


## Medical Diagnosis

- **Task ($T$)**: Classify images of skin lesions, e.g. melanoma (malignant) vs. benign nevus; the network output $y = f(\mathbf{x}; \mathbf{w})$ predicts the target $t \in \{0, 1\}$.
- **Performance ($P$)**: Sensitivity (fraction of malignant lesions detected), specificity (fraction of benign lesions correctly cleared) and the area under the ROC curve (AUC) summarizing both, measured against biopsy results *(→ Lecture 06_mlops)*.
- **Experience ($E$)**: 129,450 clinical images (3,374 of them dermoscopic) of 2,032 diseases, from 18 online repositories and Stanford University Medical Center. The network is trained on 757 disease classes; for the benign/malignant decision their probabilities are summed (figure).

<center>
    <img src="https://raw.githubusercontent.com/konstantin-schekotihin/dl_course/master/shared/images/esteva_2017_figure.jpg" width="85%"/>
</center>

- *Architecture*: Deep convolutional neural network (Inception-v3) with ~25 million adjustable weights $\mathbf{w}$, demonstrating performance on par with 21 board-certified dermatologists.

<small>📖 *Source: [Esteva, A., Kuprel, B., Novoa, R. A., et al. (2017). Dermatologist-level classification of skin cancer with deep neural networks. Nature, 542(7639), 115–118](https://doi.org/10.1038/nature21056)*</small>


## ⚡ Spotlight: Transfer Learning

- In high-stakes specialized domains like clinical healthcare, training 25 million parameters from scratch on small clinical datasets leads to severe overfitting.
- **The Transfer Learning Strategy**:
    1. **Pretraining**: Train the deep network on a large, general image collection (ImageNet: 1.28 million images of 1,000 everyday object categories) to learn general visual features (edges, textures, shapes).
    2. **Fine-tuning**: Replace the final classification layer and continue training the **whole network** with a small learning rate on the skin-lesion images, as Esteva et al. did.
- A cheaper variant freezes the pretrained layers and trains only the new final layers.

> 💡 **Key Takeaway: Transfer Learning Principle**
>
> Knowledge acquired from a data-rich source domain accelerates convergence and regularizes representation learning in a data-scarce target domain.


## Protein Structure Prediction — AlphaFold

- **Task ($T$)**: Predict the 3D position $\mathbf{r}_i \in \mathbb{R}^3$ of every atom $i$ in a protein from its 1D sequence of amino acids.
- **Performance ($P$)**: Agreement with experimentally solved structures (X-ray crystallography, cryo-EM): the Global Distance Test (GDT, fraction of atoms placed within a distance cut-off) and the root-mean-square deviation (RMSD) of the atom positions.
- **Experience ($E$)**: ~170,000 experimentally solved 3D structures deposited in the Protein Data Bank (PDB).

## Generative Image Synthesis

- **Task ($T$)**: Generate new, photorealistic images that look like samples from the training distribution: $\mathbf{x}_\mathrm{new} \sim p(\mathbf{x})$.
- **Performance ($P$)**: How close generated images are to real ones, e.g. the Fréchet Inception Distance (FID), and human ratings.
- **Experience ($E$)**: Unlabelled images, here studio photographs of faces (unsupervised learning).
- With a text **prompt** $\mathbf{c}$, the same idea generates images matching the text, $\mathbf{x} \sim p(\mathbf{x} \mid \mathbf{c})$, trained on billions of image–text pairs.

## Large Language Models

- **Task ($T$)**: Predict the next token $\mathbf{x}_n$ of a text from the previous ones:
  $$p(\mathbf{x}_n \mid \mathbf{x}_1, \dots, \mathbf{x}_{n-1})$$
- **Performance ($P$)**: Cross-entropy of the predicted tokens *(→ Lecture 03_information)* during training; benchmarks of knowledge, reasoning and coding for the finished model.
- **Experience ($E$)**: Web-scale text and source code; e.g. the open Qwen3 models were pre-trained on about 36 trillion tokens.
- *Impact*: Today's frontier models (autumn 2026: OpenAI GPT-6, Anthropic Claude 5, Google Gemini 3, and open-weight models such as Qwen and DeepSeek) write code, solve multi-step problems and follow instructions. Whether such abilities "emerge" abruptly with scale is debated; part of the effect depends on how they are measured.

<small>📖 *Source: [Vaswani, A., Shazeer, N., Parmar, N., et al. (2017). Attention is all you need. NeurIPS](https://proceedings.neurips.cc/paper/2017/hash/3f5ee243547dee91fbd053c1c4a845aa-Abstract.html)*</small><br>
<small>📖 *Source: [Qwen Team (2025). Qwen3 Technical Report. arXiv:2505.09388](https://arxiv.org/abs/2505.09388)*</small><br>
<small>📖 *Source: [Schaeffer, R., Miranda, B., & Koyejo, S. (2023). Are emergent abilities of large language models a mirage? NeurIPS](https://arxiv.org/abs/2304.15004)*</small>


# Conclusions & Key Takeaways


- **Machine Learning as Inductive Inference**: Learning generalizable models from finite training observations.
- **The Subsymbolic Triumph**: Shifting from brittle human-crafted rules to continuous neural representations optimized end-to-end.
- **The Power of Scale**: Driven by Big Data, GPU parallel compute, and scalable architectures (The "Bitter Lesson").
- **Formulating the Task**: Rigorous definition of Task $T$, Performance measure $P$, and Experience $E$ is essential for every learning project.
