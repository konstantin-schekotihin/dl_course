# Mathematical Notation

> ⚠️ **Important:** The mathematical notation used throughout this course follows **Christopher M. Bishop and Hugh Bishop (2024), _Deep Learning: Foundations and Concepts_** (Springer) and _Pattern Recognition and Machine Learning_ (2006). All formulas use standard, pure LaTeX notation without custom macro dependencies.

## Numbers and Arrays

* $x$: A scalar
* $\mathbf{x}$: A vector (assumed to be a column vector unless otherwise stated)
* $\mathbf{x}^\mathrm{T}$: Transpose of vector $\mathbf{x}$ (a row vector)
* $\mathbf{w} = (w_1, \dots, w_M)^\mathrm{T}$: Column vector with $M$ elements, where $(w_1, \dots, w_M)$ denotes the corresponding row vector
* $\mathbf{X}$ or $\mathbf{M}$: A matrix
* $\mathbf{X}^\mathrm{T}$: Transpose of matrix $\mathbf{X}$
* $\mathbf{I}_M$ or $\mathbf{I}$: The $M \times M$ identity matrix (unit matrix), with elements $I_{ij} = \delta_{ij}$ (where $\delta_{ij} = 1$ if $i = j$, and $0$ if $i \neq j$)
* $\mathbf{1}$: A column vector in which all elements have the value 1
* $x_i$: The $i^\mathrm{th}$ element of vector $\mathbf{x}$
* $X_{ij}$ or $x_{ij}$: The element of matrix $\mathbf{X}$ at row $i$ and column $j$
* $\mathbf{X} \in \mathbb{R}^{N \times D}$: A data matrix of $N$ observations, each of dimension $D$, where the $n^\mathrm{th}$ row corresponds to observation $\mathbf{x}_n^\mathrm{T}$ and element $(n, i)$ is $x_{ni}$
* $\mathbf{x} = (x_1, \dots, x_N)^\mathrm{T}$: A data column vector combining $N$ one-dimensional observations (distinguished from a $D$-dimensional feature vector $\mathbf{x}$ by dimension and context)
* $\mathbf{X}$ or $X$ with elements $x_{ijkn}$: A higher-dimensional array (tensor) with multiple indices (e.g., height $i$, width $j$, channel $k$, image $n$)
* $\mathrm{diag}(\mathbf{a})$: A square, diagonal matrix with diagonal entries given by vector $\mathbf{a}$
* $t$: Target variable (scalar observation)
* $\mathbf{t} \in \mathbb{R}^K$: Target vector (e.g. 1-of-$K$ one-hot classification vector $\mathbf{t} \in \{0, 1\}^K$ or multi-target continuous vector)
* $\mathbf{t} = (t_1, \dots, t_N)^\mathrm{T}$: Target column vector combining $N$ one-dimensional target observations
* $\mathbf{T} \in \mathbb{R}^{N \times K}$: Target data matrix combining $N$ target vectors
* $\mathcal{D} = \{(\mathbf{x}_n, \mathbf{t}_n)\}_{n=1}^N$: Supervised training dataset consisting of $N$ input-target pairs
* $x \sim p(x)$, $\mathbf{x} \sim p(\mathbf{x})$: Random variables (random variables share standard variable typography; context distinguishes random variables from specific realizations)

## Sets and Logic

* $\mathcal{S}$ or $\mathcal{X}$: A set
* $\{x_1, \dots, x_N\}$: A set defined by explicit enumeration
* $\forall m \in \mathcal{M}$: Universal quantifier ("for all $m$ in set $\mathcal{M}$ ")
* $\mathbb{N}$: The set of natural numbers $\{1, 2, 3, \dots\}$
* $\mathbb{Z}$: The set of integers
* $\mathbb{Z}^+$: The set of positive integers
* $\mathbb{R}$: The set of real numbers
* $\mathbb{R}^D$: The space of $D$-dimensional real column vectors
* $\mathbb{R}^{N \times D}$: The space of $N \times D$ real matrices
* $|\mathcal{S}|$: Cardinality (number of elements) of set $\mathcal{S}$
* $\mathcal{A} \cup \mathcal{B}$: Union of sets $\mathcal{A}$ and $\mathcal{B}$
* $\mathcal{A} \cap \mathcal{B}$: Intersection of sets $\mathcal{A}$ and $\mathcal{B}$
* $\mathcal{A} \setminus \mathcal{B}$: Set difference (elements in $\mathcal{A}$ that are not in $\mathcal{B}$)
* $\mathcal{N}(i)$: The set of neighbors of node $i$ on a graph (distinguished by context from the Gaussian distribution $\mathcal{N}$)

## Functions and Neural Network Operations

* $f(\cdot)$ or $f: \mathcal{A} \to \mathcal{B}$: A function $f$ mapping domain $\mathcal{A}$ to range $\mathcal{B}$
* $y(\mathbf{x}, \mathbf{w})$ or $f(\mathbf{x}; \mathbf{w})$: A model function of input $\mathbf{x}$ parametrized by weight vector $\mathbf{w}$
* $f[y]$: A functional (a mapping taking an entire function $y(x)$ as argument to return a scalar value)
* $\ln(\cdot)$ or $\log(\cdot)$: Natural logarithm (base $e$)
* $\exp(\cdot)$: The exponential function, $\exp(x) = e^x$
* $f \circ g$: Composition of functions, $(f \circ g)(x) = f(g(x))$
* $\sigma(a)$: Logistic sigmoid function, $\sigma(a) = \frac{1}{1 + \exp(-a)}$
* $\operatorname{softmax}(\mathbf{a})$: Softmax activation function with elements $\frac{\exp(a_k)}{\sum_j \exp(a_j)}$
* $\max(0, a)$: Rectified Linear Unit (ReLU) activation function
* $\mathbb{I}(\text{condition})$: The indicator function (evaluates to $1$ if the condition is true, and $0$ otherwise)
* $|x|$: The modulus (absolute value / positive part) of scalar $x$
* $\lfloor x \rfloor$: The floor of $x$ (the largest integer less than or equal to $x$)

## Operators and Linear Algebra

* $\mathbf{A}^\mathrm{T}$: Transpose of matrix $\mathbf{A}$, satisfying $(\mathbf{A}\mathbf{B})^\mathrm{T} = \mathbf{B}^\mathrm{T}\mathbf{A}^\mathrm{T}$
* $\mathbf{A}^{-1}$: Inverse of matrix $\mathbf{A}$, satisfying $\mathbf{A}\mathbf{A}^{-1} = \mathbf{A}^{-1}\mathbf{A} = \mathbf{I}$
* $\mathbf{a} \oplus \mathbf{b}$: Concatenation of vectors $\mathbf{a} = (a_1, \dots, a_N)^\mathrm{T}$ and $\mathbf{b} = (b_1, \dots, b_M)^\mathrm{T}$ into $(a_1, \dots, a_N, b_1, \dots, b_M)^\mathrm{T}$
* $\mathbf{x}^\mathrm{T}\mathbf{y}$: Inner product (dot product) of column vectors $\mathbf{x}$ and $\mathbf{y}$
* $\mathbf{A} \odot \mathbf{B}$: Hadamard (elementwise) product of matrices $\mathbf{A}$ and $\mathbf{B}$
* $\|\mathbf{x}\|$: Euclidean ($L_2$) norm of vector $\mathbf{x}$, $\|\mathbf{x}\| = (\mathbf{x}^\mathrm{T}\mathbf{x})^{1/2} = \sqrt{\sum_i x_i^2}$
* $\|\mathbf{x}\|_p$: $L_p$ norm, $\|\mathbf{x}\|_p = \left(\sum_i |x_i|^p\right)^{1/p}$
* $\det \mathbf{A}$ or $|\mathbf{A}|$: Determinant of square matrix $\mathbf{A}$
* $\mathrm{Tr}(\mathbf{A})$: Trace of square matrix $\mathbf{A}$ (sum of diagonal entries, $\sum_i A_{ii}$)
* $\mathrm{Tr}(\mathbf{ABC}) = \mathrm{Tr}(\mathbf{CAB}) = \mathrm{Tr}(\mathbf{BCA})$: Cyclic property of the trace operator
* $\sum$: Summation operator
* $\prod$: Product operator
* $\triangleq$ or $\stackrel{\mathrm{def}}{=}$: Equal by definition

## Calculus and Matrix Derivatives

* $\frac{df}{dx}$: Derivative of scalar function $f$ with respect to scalar $x$
* $\frac{\partial f}{\partial x}$: Partial derivative of $f$ with respect to $x$
* $\nabla_{\mathbf{x}} y$ or $\frac{\partial y}{\partial \mathbf{x}}$: Gradient of scalar $y$ with respect to column vector $\mathbf{x}$, with elements $\left(\frac{\partial y}{\partial \mathbf{x}}\right)_i = \frac{\partial y}{\partial x_i}$
* $\frac{\partial \mathbf{a}}{\partial x}$: Derivative of vector $\mathbf{a}$ with respect to scalar $x$, with elements $\left(\frac{\partial \mathbf{a}}{\partial x}\right)_i = \frac{\partial a_i}{\partial x}$
* $\frac{\partial \mathbf{a}}{\partial \mathbf{b}}$: Jacobian matrix of vector $\mathbf{a}$ with respect to vector $\mathbf{b}$, with elements $\left(\frac{\partial \mathbf{a}}{\partial \mathbf{b}}\right)_{ij} = \frac{\partial a_i}{\partial b_j}$
* $\frac{\partial}{\partial \mathbf{x}}(\mathbf{x}^\mathrm{T}\mathbf{a}) = \frac{\partial}{\partial \mathbf{x}}(\mathbf{a}^\mathrm{T}\mathbf{x}) = \mathbf{a}$: Derivative of linear form
* $\frac{\partial}{\partial \mathbf{A}}\mathrm{Tr}(\mathbf{AB}) = \mathbf{B}^\mathrm{T}$, $\frac{\partial}{\partial \mathbf{A}}\mathrm{Tr}(\mathbf{A}) = \mathbf{I}$: Matrix trace derivatives
* $\frac{\partial}{\partial \mathbf{A}}\ln |\mathbf{A}| = (\mathbf{A}^{-1})^\mathrm{T}$: Derivative of log determinant
* $\int_a^b f(x)\,dx$: Definite integral of $f(x)$ from $a$ to $b$
* $\int f(x)\,dx$: Indefinite integral with respect to $x$

## Probability and Information Theory

* $p(X = x_i)$ or $P(X = x)$: Formal probability mass that discrete random variable $X$ takes realization value $x_i \in \mathcal{X}$
* $p(x)$: Probability distribution (PMF) or probability density (PDF) over variable $x$ (to avoid cumbersome notation, random variables share standard lowercase typography $x$, with context distinguishing the variable from a realization; uppercase bold $\mathbf{X}$ is reserved for data matrices)
* $p(\mathbf{x})$: Joint probability distribution or density over vector $\mathbf{x}$
* $p(y \mid x)$: Conditional distribution of $y$ given $x$
* $p(x, y)$: Joint distribution of $x$ and $y$
* $x \sim p(x)$: Variable $x$ is distributed according to (sampled from) distribution $p(x)$
* $p_x(\cdot)$: Density notation with explicit variable subscript to avoid ambiguity
* $\mathbb{E}_x[f(x, y)]$ or $\mathbb{E}[f(x)]$: Expectation of $f(x)$ with respect to distribution $p(x)$
* $\mathbb{E}_x[f(x) \mid z]$: Conditional expectation of $f(x)$ given $z$
* $\mathrm{var}[f(x)]$ or $\mathrm{var}[x]$: Variance of $f(x)$ under $p(x)$, $\mathrm{var}[x] = \mathbb{E}[(x - \mathbb{E}[x])^2]$
* $\mathrm{cov}[\mathbf{x}, \mathbf{y}]$: Covariance matrix of vectors $\mathbf{x}$ and $\mathbf{y}$, $\mathrm{cov}[\mathbf{x}, \mathbf{y}] = \mathbb{E}[(\mathbf{x} - \mathbb{E}[\mathbf{x}])(\mathbf{y} - \mathbb{E}[\mathbf{y}])^\mathrm{T}]$
* $\mathrm{cov}[\mathbf{x}] = \mathrm{cov}[\mathbf{x}, \mathbf{x}]$: Covariance matrix of vector $\mathbf{x}$
* $\mathcal{N}(\mathbf{x} \mid \boldsymbol{\mu}, \mathbf{\Sigma})$: Multivariate Gaussian (Normal) distribution with mean vector $\boldsymbol{\mu}$ and covariance matrix $\mathbf{\Sigma}$
* $\mathbf{\Lambda} = \mathbf{\Sigma}^{-1}$: Precision matrix (inverse covariance matrix); $\beta = 1/\sigma^2$ for scalar precision
* $x \perp y$: Variables $x$ and $y$ are statistically independent, $p(x, y) = p(x)p(y)$
* $x \perp y \mid z$: Variables $x$ and $y$ are conditionally independent given $z$, $p(x, y \mid z) = p(x \mid z)p(y \mid z)$
* $\mathrm{H}[x]$ or $H(x)$: Entropy of variable $x$
* $\mathrm{KL}(p \parallel q)$ or $D_\mathrm{KL}(p \parallel q)$: Kullback–Leibler divergence between distributions $p$ and $q$

## Asymptotic Complexity

* $\mathcal{O}(f(x))$: Big $\mathcal{O}$ notation ($g(x) = \mathcal{O}(f(x))$ denotes that $|g(x)/f(x)|$ remains bounded as $x \to \infty$)
