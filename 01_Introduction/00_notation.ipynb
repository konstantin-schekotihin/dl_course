# Notation

<div class="myalert">The notation used throughout this course is summarized below</div>

## Numbers and Arrays

* $x$: A scalar
* $\vec{x}$: A vector
* $\vec{X}$: A matrix
* $\tens{X}$: A tensor
* $\vec{I}_n$: An identity matrix with $n$ rows and columns, or simply $\vec{I}$ if clear from the context
* $x_i$, $[\vec{x}]_i$: The $i^\mathrm{th}$ element of vector $\vec{x}$
* $x_{ij}$, $x_{i,j}$,$[\vec{X}]_{ij}$, $[\vec{X}]_{i,j}$: The element of matrix $\vec{X}$ at row $i$ and column $j$
* $\vec{A}_{i,:}$: Row $i$ of matrix $\vec{A}$
* $\vec{A}_{:,i}$: Column $i$ of matrix $\vec{A}$
* $\tensel{A}_{i,j,k}$: Element $(i,j,k)$ of a 3-D tensor $\tens{A}$ 
* $\tens{A}_{:,:,i}$ : 2-D slice of a 3-D tensor
* $\diag{a}$: A square, diagonal matrix with diagonal entries given by $\vec{a}$
* $\rvar{a}$: A scalar random variable 
* $\rvec{a}$: A vector-valued random variable
* $\rvec{A}$: A matrix-valued random variable

## Set Theory

* $\st{X}$: A set
* $\mathbb{N}$: The set of natural numbers
* $\mathbb{Z}$: The set of integers
* $\mathbb{Z}^+$: The set of positive integers
* $\mathbb{R}$: The set of real numbers
* $\mathbb{R}^n$: The set of $n$-dimensional vectors of real numbers
* $\mathbb{R}^{a\times b}$: The set of matrices of real numbers with $a$ rows and $b$ columns
* $|\st{X}|$: Cardinality (number of elements) of set $\st{X}$
* $\st{A}\cup\st{B}$: Union of sets $\st{A}$ and $\st{B}$
* $\st{A}\cap\st{B}$: Intersection of sets $\st{A}$ and $\st{B}$
* $\st{A}\setminus\st{B}$: Subtraction of set $\st{B}$ from set $\st{A}$

## Functions

* $f(\cdot)$ or $f : \st{A} \to \st{B}$: The function $f$ with domain $\st{A}$ and range $\st{B}$
* $\log(\cdot)$: The natural logarithm
* $\exp(\cdot)$: The exponential function
* $f(\vec{x}; \vec{\theta})$: a function of $\vec{x}$ parametrized by $\vec{\theta}$
* $f \circ g$: Composition of the functions $f$ and $g$
* $\mathbf{1}_\text{condition}$: The indicator function that returns 1 if the condition is true, 0 otherwise
* $\sigma(x)$: Logistic sigmoid function $\sigma(x) = \frac{1}{1+\exp(-x)}$
* $\zeta(x)$: Softplus function $\zeta(x) = \log(1+\exp(x))$
* $x^+$: Positive part of $x$, i.e., $x^+=\max(0,x)$


## Operators

* $\mathbf{(\cdot)}^\top$: Transpose of a vector or a matrix
* $\mathbf{X}^{-1}$: Inverse of matrix $\mathbf{X}$
* $\vec{A}\odot\vec{B}$: Hadamard (elementwise) product of $\vec{A}$ and $\vec{B}$
* $[\cdot, \cdot]$: Concatenation
* $\|\cdot\|_p$: $L_p$ norm
* $\|\cdot\|$: $L_2$ norm
* $\langle \mathbf{x}, \mathbf{y} \rangle$: Dot product of vectors $\mathbf{x}$ and $\mathbf{y}$
* $\sum$: Series addition
* $\prod$: Series multiplication
* $\stackrel{\mathrm{def}}{=}$: Definition

## Calculus

* $\frac{df}{dx}$: Derivative of $f$ with respect to $x$
* $\frac{\partial f}{\partial x}$: Partial derivative of $f$ with respect to $x$
* $\nabla_{\vec{x}} y$: Gradient of $y$ with respect to $\vec{x}$
* $\int_a^b f(x) \;dx$: Definite integral of $f$ from $a$ to $b$ with respect to $x$
* $\int f(x) \;dx$: Indefinite integral of $f$ with respect to $x$

## Probability and Information Theory

* $P(\cdot)$: Probability distribution over a discrete variable
* $p(\cdot)$: Probability distribution over a continuous variable
* $z \sim P$: Random variable $z$ has probability distribution $P$
* $\mathbb{E}_{\rvar{x} \sim P} [f(x)]$ or $\mathbb{E}[f(x)]$: Expectation of $f(x)$ with respect to $P(\rvar{x})$
* $\rvar{x} \perp \rvar{y}$: Random variables $\rvar{x}$ and $\rvar{y}$ are independent
* $\rvar{x} \perp \rvar{y} \mid \rvar{z}$: Random variables $\rvar{x}$ and $\rvar{x}$ are conditionally independent given random variable $\rvar{z}$
* $\mathrm{Var}(f(x))$: Variance of $f(x)$ under $P(\rvar{x})$
* $\sigma_{\rvar{x}}$: Standard deviation of random variable $\rvar{x}$
* $\mathrm{Cov}(\rvar{x}, \rvar{y})$: Covariance of $\rvar{x}$ and $\rvar{y}$
* $\mathrm{Var}(\rvar{x}) = \mathrm{Cov}(\rvar{x}, \rvar{x})$: Variance of $\rvar{x}$
* $\rho_{\rvar{x}, \rvar{y}}$: Correlation of random variables $\rvar{x}$ and $\rvar{y}$
* $H(\rvar{x})$: Entropy of random variable $\rvar{x}$
* $D_{\mathrm{KL}}(P\|Q)$: KL-divergence of distributions $P$ and $Q$

## Complexity

* $\mathcal{O}$: Big O notation
